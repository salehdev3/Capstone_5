```python
from google.colab import drive
from pathlib import Path
import hashlib, urllib.request, importlib.util

drive.mount('/content/drive')
ROOT = Path('/content/drive/MyDrive/CameraTrap_Phase2')
script = Path('/content/export_detection_size_breakdown.py')
with urllib.request.urlopen('https://raw.githubusercontent.com/salehdev3/Capstone_5/phase-2/tools/export_detection_size_breakdown.py', timeout=45) as response:
    source = response.read()
assert hashlib.sha256(source).hexdigest() == '5d3df2e7928b357e368ab6930438bab8220aff3fa3f924c6c0173303a643f923'
script.write_bytes(source)
spec = importlib.util.spec_from_file_location('size_breakdown', script)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
SIZE_OUTPUT = module.run(ROOT)
```
